```python
def barrett_reduction(x, q, mu, logq, k, debug=True):
    assert x < (q-1)*(q-1)
    assert x < (1 << (2*logq))
    ch = x >> logq
    print("ch", hex(ch))
    t0 = ch * mu
    print("t0", hex(t0))
    t0h = t0 >> logq
    print("t0h", hex(t0h))
    t1 = t0h * q
    print("t1", hex(t1))
    cl = x & ((1 << (k+2)) - 1)
    print("cl", hex(cl))
    t2 = (cl - t1) & ((1 << (logq+2)) - 1)
    print("t2", hex(t2))
    print("t1''", hex(-t2 & 0xffffffff))
    t3 = t2 - q
    print("t3", hex(t3))
    t4 = t3 - q
    print("t4", hex(t4))
    if (t2 >= q):
        if (t3 >= q):
            return t4
        else:
            return t3
    else:
        return t2
```

```python
from tqdm.auto import tqdm
q    = 12289
logq = 13
mu = (1 << (2* logq)) // q

for i in tqdm(range(q**2)):
    if (barrett_reduction(i, q, mu, logq) != i % q):
        print(i, barrett_reduction(i, q, mu, logq))
        break
```

```python
#### ADDRESS_GEN WORK BELOW



logn = 4
n = 1 << logn

bram_0 = [i for i in range(n//2)]
bram_1 = [i + (n//2) for i in range(n//2)]


print("bram_0: ", bram_0)
print("bram_1: ", bram_1)


for j in range(logn):
    bu = []
    for i in range(n//2):
        mask = (1 << ((logn - j - 1))) - 1 if (j != (logn - 1)) else 1
        shift = (logn - j - 2) if (j != (logn - 1)) else 0
        addr = ((i & 0x1) << (shift)) + ((i & mask) >> 1) + (i & ~mask)
        data = (bram_0[addr], bram_1[addr])
        print(f"stage: {j}, i: {i}, addr: {addr}, mask: {mask:0{logn}b}, shift:{shift}, data: {data}")
        bu.append(data)
    print("bu: ", bu)

    for i in range(n//2):

        mask = (1 << ((logn - j - 1))) - 1 if (j != (logn - 1)) else 1
        shift = (logn - j - 2) if (j != (logn - 1)) else 0
        addr_0 = ((i & 0x1) << (shift)) + ((i & mask) >> 1) + (i & ~mask)



        out_even = bu[i][0]


        if j != logn - 1:
            flag = i & 0x1

            if flag == 0:
                bram_0[addr_0] = out_even
            else:
                bram_1[addr_1] = out_even

            print(f"i: {i}, addr_0: {addr_0}, out_even: {out_even}, flag: {flag}")

            if i > 0:
                out_odd  = bu[i - 1][1]
                if flag == 0:
                    bram_1[addr_1] = out_odd
                else:
                    bram_0[addr_0] = out_odd
                print(f"i: {i}, addr_1: {addr_1}, out_odd: {out_odd}, flag: {flag}")
            addr_1 = addr_0
        else:
            out_odd  = bu[i][1]
            bram_0[addr_0] = out_even
            bram_1[addr_0] = out_odd
            print(f"i: {i}, addr_0: {addr_0}, out_even: {out_even}, out_odd: {out_odd}")


        print("bram_0: ", bram_0)
        print("bram_1: ", bram_1)
```
