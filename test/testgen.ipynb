```python
def barrett_reduction(x, q, mu, logq, k, debug=True):
    assert x < (q-1)*(q-1)
    assert x < (1 << (2*logq))
    ch = x >> logq
    print("ch", hex(ch))
    t0 = ch * mu
    print("t0", hex(t0))
    t0h = t0 >> logq
    print("t0h", hex(t0h))
    t1 = t0h * q
    print("t1", hex(t1))
    cl = x & ((1 << (k+2)) - 1)
    print("cl", hex(cl))
    t2 = (cl - t1) & ((1 << (logq+2)) - 1)
    print("t2", hex(t2))
    print("t1''", hex(-t2 & 0xffffffff))
    t3 = t2 - q
    print("t3", hex(t3))
    t4 = t3 - q
    print("t4", hex(t4))
    if (t2 >= q):
        if (t3 >= q):
            return t4
        else:
            return t3
    else:
        return t2
```

```python
from tqdm.auto import tqdm
q    = 12289
logq = 13
mu = (1 << (2* logq)) // q

for i in tqdm(range(q**2)):
    if (barrett_reduction(i, q, mu, logq) != i % q):
        print(i, barrett_reduction(i, q, mu, logq))
        break
```
